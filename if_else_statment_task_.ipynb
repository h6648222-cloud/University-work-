```python
# 5-10 Checking Usernames

current_users = ["hamza", "admin", "jaden", "sara", "alex"]
new_users = ["Hamza", "ali", "john", "sara", "zain"]

# create lowercase version of current users
current_users_lower = []

for user in current_users:
    current_users_lower.append(user.lower())

for new_user in new_users:
    if new_user.lower() in current_users_lower:
        print(new_user + " is already taken. Please choose another username.")
    else:
        print(new_user + " is available.")
```
