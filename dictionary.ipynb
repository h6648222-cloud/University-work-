```python
# if else condition in dictionary
glossary = {
    "name": "A word used to identify a person.",
    "age": "How old a person is.",
    "city": "The place where a person lives.",
    "phone": "A device used for calling and texting.",
    "book": "Something we read to get knowledge.",
    "school": "A place where students study.",
    "teacher": "A person who teaches students.",
    "student": "A person who learns in school or university.",
    "class": "A group or lesson in school.",
    "bag": "Something used to carry books and things."
}

for key, value in glossary.items():
    print(key , ":")
    print(" ", value)
    print()
```

```python
rivers = {
    "nile": "egypt",
    "indus": "pakistan",
    "amazon": "brazil"
}

# Print a sentence about each river
for river, country in rivers.items():
    print(f'The {river.title()} runs through {country.title()}.')

print("\nRivers:")
# Print the name of each river
for river in rivers.keys():
    print(river.title())

print("\nCountries:")
# Print the name of each country
for country in rivers.values():
    print(country.title())
```

```python
# Existing dictionary of favorite languages
favorite_languages = {
    "alice": "python",
    "bob": "c",
    "carol": "java"
}

# List of people who should take the poll
people_to_poll = ["alice", "david", "bob", "emma", "frank"]

# Loop through the list and check
for person in people_to_poll:
    if person in favorite_languages:
        print("Thank you, " , person.title() , ", for responding to the poll.")
    else:
        print("Hi " , person.title() , ", please take the poll!")
```

```python
# Dictionary in Dictionary
users = {
    'aeinstein': {
        'first': 'albert',
        'last': 'einstein',
        'location': 'princeton',
    },
    'mcurie': {
        'first': 'marie',
        'last': 'curie',
        'location': 'paris',
    },
}
for username, user_info in users.items():
    print(f"\nUsername: {username}")
    full_name = f"{user_info['first']} {user_info['last']}"
    location = user_info['location']
    print(f"\tFull name: {full_name.title()}")
    print(f"\tLocation: {location.title()}")
```

```python
# Three people dictionaries
person1 = {"first_name": "Alice", "last_name": "Smith", "age": 25, "city": "New York"}
person2 = {"first_name": "Bob", "last_name": "Brown", "age": 30, "city": "Los Angeles"}
person3 = {"first_name": "Carol", "last_name": "Davis", "age": 22, "city": "Chicago"}

# Store them in a list
people = [person1, person2, person3]

# Loop through list and print info
for person in people:
    print("\nPerson Info:")
    for key, value in person.items():
        print(key.title() , ":", value)
```

```python
# Dictionaries for each pet
pet1 = {"name": "Max", "animal": "dog", "owner": "Alice"}
pet2 = {"name": "Whiskers", "animal": "cat", "owner": "Bob"}
pet3 = {"name": "Tweety", "animal": "bird", "owner": "Carol"}

# Store in a list
pets = [pet1, pet2, pet3]

# Loop and print
for pet in pets:
    print("\nPet Info:")
    for key, value in pet.items():
        print(key.title() ,":", value)
```

```python
favorite_places = {
    "Ali": ["Paris", "Tokyo"],
    "AShwariya": ["New York", "London", "Sydney"],
    "KAtrina": ["Rome"]
}

# Loop through dictionary
for person, places in favorite_places.items():
    print("\n" + person + "'s favorite places are:")
    for place in places:
        print("-", place)
```
